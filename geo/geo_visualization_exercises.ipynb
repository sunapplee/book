# Практика: геоданные и графики

Цель — освежить работу с векторными и растровыми геоданными, статическими графиками и интерактивной визуализацией.

Базовый блок — задания 1–6, два подхода по 60–90 минут. Дополнительный блок — задания 7–11, ещё два подхода примерно по часу. Выполняй задания последовательно в этом ноутбуке. Сначала пробуй по памяти, затем при необходимости обращайся к справочнику book.

Для каждого задания подготовь результат и короткий вывод: что получилось, что проверил и какие ограничения заметил. В ноутбуке нет решений и заготовок кода.

## Данные

Все исходные файлы уже находятся в локальной директории champs. Пути ниже указаны относительно папки этого ноутбука — book/geo.

- Климат: ../../atomskills2026_solution/data/climate_1995_2026.csv. Для начала используй первые 100 000 строк. Основные поля: date, lat, lon, temp, precipitation.
- Границы муниципалитетов: ../../atomskills2026_solution/data/Krasnodar_Municipalities_GeoJson.txt. Файл содержит GeoJSON, несмотря на расширение TXT.
- Карта высот: ../../as26/data/elev/EarthEnv-DEM90_N45E030.bil. Сопутствующие файлы оставь рядом с растром.

Климат и границы относятся к Краснодарскому краю. DEM используется в отдельном задании: выбирай точки внутри его собственного покрытия.

Перед расчётами проверь структуру, типы данных, временной диапазон и пропуски. Не требуется загружать все климатические данные за все годы.

## 1. Точки и границы

Ориентир: 15 минут.

### Задание

1. Загрузи границы муниципалитетов и климатические наблюдения.
2. Создай пространственный слой уникальных точек наблюдений на основе координат.

4. Добавь название и различимые обозначения слоёв.

### Критерии готовности

- Проверены системы координат обоих слоёв и их совместимость.
- Долгота и широта не перепутаны.
- Одна точка не дублируется на карте для каждой даты наблюдения.
- Точки и границы отображаются в ожидаемом месте.

### Вопросы для самопроверки

## 11. Найти и исправить ошибки визуализации

Ориентир: 25 минут.

### Задание

Проведи аудит следующих ситуаций. Для каждой опиши симптом, способ проверки и исправление. На собственных данных воспроизведи ошибочный и исправленный вариант, где это возможно.

1. При создании метеоточек широту приняли за горизонтальную координату, а долготу — за вертикальную. Точки не совпадают с границами региона.
2. Координаты в градусах объявили метрическими простым назначением другой CRS. Затем построили буфер радиусом 10 000 единиц.
3. Две карты разных дат имеют независимые автоматические цветовые шкалы. По одинаковому цвету сделан вывод об одинаковой температуре.
4. Временной ряд построили по строкам в случайном порядке. Линия скачет между датами.
5. Районы сравнили по сумме осадков всех точек без учёта разного числа метеоточек.
6. Значение NoData включили в среднюю высоту и гистограмму DEM как обычное наблюдение.

### Критерии готовности

- Для всех шести случаев есть объяснение причины, проверка и исправление.
- Подготовлены сравнения «до / после» минимум для четырёх случаев.
- Исправление проверено числами или метаданными, а не только внешним видом рисунка.
- Для восстановления ошибочно назначенной CRS использована известная исходная система координат.
- В выводе перечислены проверки, которые ты будешь выполнять перед сдачей графиков и карт.

- Может ли график выглядеть убедительно и при этом показывать некорректный результат?
- Какие ошибки обнаруживаются по CRS и диапазонам координат ещё до построения карты?
- Какие три проверки ты сделаешь первыми, если точки внезапно оказались далеко от нужного региона?
